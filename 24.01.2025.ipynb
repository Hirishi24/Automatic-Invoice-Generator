```python
text = "ccHello, Python!cc"

# 1. lstrip() - Removes leading spaces (left side)
print(text.lstrip('c'))  # Output: "Hello, Python!  "

# 2. rstrip() - Removes trailing spaces (right side)
print(text.rstrip('c'))  # Output: "  Hello, Python!"

# 3. strip() - Removes spaces from both sides
print(text.strip('c'))  # Output: "Hello, Python!"
```
